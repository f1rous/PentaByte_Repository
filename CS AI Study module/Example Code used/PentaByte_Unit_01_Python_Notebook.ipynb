**2.2 — Modify a condition, a loop, and a short function**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
def count_passing(scores, pass_mark):
    count = 0
    for score in scores:
        if score >= pass_mark:
            count += 1
    return count

print(count_passing([72, 85, 68, 91, 85], 60))  # 5
print(count_passing([72, 85, 68, 91, 85], 80))  # 3
```

**2.2 — Modify a condition, a loop, and a short function**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
def find_highest_score(scores):
    if not scores:
        return None

    highest = scores[0]
    for index in range(1, len(scores)):
        if scores[index] > highest:
            highest = scores[index]
    return highest

result = find_highest_score([72, 85, 68, 91, 85])
if result is None:
    print("No scores available")
else:
    print("Highest score:", result)
```

**2.3 — Complete the score-summary program and test other pass marks**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
def count_passing(scores, pass_mark):
    total = 0
    for score in scores:
        if score >= pass_mark:
            total += 1
    return total

scores = [72, 85, 68, 91, 85]

for mark in [60, 80]:
    print("Pass mark:", mark,
          "| Students meeting it:", count_passing(scores, mark))
```

## Chapter 3 — Measuring Algorithm Efficiency

**3.2 — Recognize O(1), O(n), and O(n²) visually**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
# O(1) work: one indexed access, assuming scores is nonempty
first_score = scores[0]

# O(n) work: one pass through all scores
for score in scores:
    print(score)

# O(n^2) work: every score is paired with every score
for first in scores:
    for second in scores:
        print(first, second)
```

**3.3 — Compare keeping one maximum with copying a list: space complexity**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
def highest_without_copy(scores):
    if not scores:
        return None

    highest = scores[0]
    for index in range(1, len(scores)):
        if scores[index] > highest:
            highest = scores[index]
    return highest
```

**3.3 — Compare keeping one maximum with copying a list: space complexity**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
def highest_with_copy(scores):
    if not scores:
        return None

    copied_scores = scores.copy()
    highest = copied_scores[0]
    for index in range(1, len(copied_scores)):
        if copied_scores[index] > highest:
            highest = copied_scores[index]
    return highest
```
