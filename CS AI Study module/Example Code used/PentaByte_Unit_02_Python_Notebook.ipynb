## Chapter 5 — Stacks, Queues, Dictionaries, and Sets

**5.1 — Compare stacks and queues; recognize linked nodes**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
from collections import deque

# Stack: last in, first out
undo_stack = []
for task in ["A", "B", "C"]:
    undo_stack.append(task)

stack_order = []
while undo_stack:
    stack_order.append(undo_stack.pop())

# Queue: first in, first out
help_queue = deque(["A", "B", "C"])
queue_order = []
while help_queue:
    queue_order.append(help_queue.popleft())

print("Stack removal:", stack_order)
print("Queue removal:", queue_order)
```

## Chapter 7 — Searching and Elementary Sorting

**7.1 — Trace linear search and binary search**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
def linear_search(values, target):
    for index, value in enumerate(values):
        if value == target:
            return index
    return -1


def binary_search(sorted_values, target):
    left = 0
    right = len(sorted_values) - 1

    while left <= right:
        middle = (left + right) // 2
        value = sorted_values[middle]

        if value == target:
            return middle
        if value < target:
            left = middle + 1
        else:
            right = middle - 1

    return -1


scores = [68, 72, 85, 91, 96]
print("Linear index:", linear_search(scores, 85))
print("Binary index:", binary_search(scores, 85))
print("Missing item:", binary_search(scores, 99))
```

## Chapter 8 — Graph Traversal and Shortest Paths

**8.2 — Trace a supplied BFS example and reconstruct the route**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
from collections import deque

campus = {
    "Dorm": ["Library", "Canteen"],
    "Library": ["Dorm", "Lab"],
    "Canteen": ["Dorm", "Gate"],
    "Lab": ["Library", "Gate"],
    "Gate": ["Canteen", "Lab"],
}


def shortest_path(graph, start, goal):
    """Return a fewest-edge path, or None if no path exists."""
    if start not in graph or goal not in graph:
        return None

    queue = deque([start])
    visited = {start}
    parent = {start: None}

    while queue:
        current = queue.popleft()
        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            return list(reversed(path))

        for neighbor in graph[current]:
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current
                queue.append(neighbor)

    return None


route = shortest_path(campus, "Dorm", "Lab")
print("Route:", " -> ".join(route) if route else "Unreachable")
print("Number of edges:", len(route) - 1 if route else "N/A")
```

**8.3 — Change a connection, test an unreachable destination, and interpret the result**

Run the cell below, inspect its output, and compare it with the corresponding lesson in the Markdown document.

```python
# Make a separate graph so the original campus map remains unchanged.
blocked_campus = {
    location: neighbors.copy()
    for location, neighbors in campus.items()
}

# Remove both undirected connections to the Lab.
for neighbor in ["Library", "Gate"]:
    blocked_campus[neighbor].remove("Lab")
    blocked_campus["Lab"].remove(neighbor)

blocked_route = shortest_path(blocked_campus, "Dorm", "Lab")
print("Route after closures:", blocked_route)
print("Lab neighbors:", blocked_campus["Lab"])
```
